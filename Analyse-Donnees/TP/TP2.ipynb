Question 3

```python
df1 = df.dropna()
```

```python
df2 = df.copy()

for col in df2.select_dtypes(include=np.number).columns:
    df2[col] = df2[col].fillna(df2[col].mean())
```

```python
df3 = df.copy()

for col in df3.select_dtypes(include=np.number).columns:
    df3[col] = df3[col].fillna(df3[col].median())
```

```python
for col in df3.select_dtypes(exclude=np.number).columns:
    if not df3[col].mode().empty:
        df3[col] = df3[col].fillna(df3[col].mode()[0])
```

Question 4

```python
df3["Volume"] = pd.to_numeric(df3["Volume"], errors="coerce")
df3["Volume"] = df3["Volume"].fillna(df3["Volume"].median())

vmin = df3["Volume"].min()
vmax = df3["Volume"].max()

if vmax != vmin:
    df3["Volume"] = (df3["Volume"] - vmin) / (vmax - vmin)
else:
    df3["Volume"] = 0
```

Question 5

```python
print("Doublons avant :", df3.duplicated().sum())

df3 = df3.drop_duplicates()

print("Doublons après :", df3.duplicated().sum())
```

Question 6

```python
df3.to_csv("out_data.csv", index=False)
```
